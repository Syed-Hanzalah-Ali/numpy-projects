# Project 2- Airport Cabs Data Analysis

```python
import numpy as np
```

```python
data=np.genfromtxt('nyc_taxis.csv',delimiter=',',skip_header=True)  # data would be in float type
print(data)
```

### Mean speed of all the cab rides

```python
# trip length is time
timeInHours=data[:,8]/3600   # seconds to hours
# print(timeInHours)

distance=data[:,7]
# print(distance)

# speed=distance/time(hours)
speed=distance/timeInHours
# print(speed)   # km/h

speedMean=np.mean(speed)
print("\nmean of speed is:\n",speedMean)
```

### Calculate the number of rides taken in any specific month

```python
# pickup_month is from 1..12, denoting jan .. dec respectively

# 2nd col is month column
ridesInFebruary=data[data[:,1]==2]
print(ridesInFebruary.shape)   # gives (rows,col)
print("\nNumber of rides taken in the month of February:\n",ridesInFebruary.shape[0])
```

### Calculate the number of rides with a tip greater than $50

```python
# 12th col is tim_Amount column, -3 if count from last
tipAmounts=data[data[:,-3]>50,-3]  # data[condition]=> select all column. 2nd argument for 12th column

print("\nNumber of rides with a tip greater than 50$ are:\n",tipAmounts.shape[0])
```

### Calculate the number of rides where drop was JFK airport

```python
# drop of location code for JFK is assume 2
dropOffAtJFK=data[data[:,6]==2,6]
print("\nNumber of rides where dropOff was JFK airport are:\n",dropOffAtJFK.shape[0])
```
