# Company's sales Data of 10 years

```python
import numpy as np
```

```python
data=np.loadtxt("random_unsorted_sales_data.csv",dtype=str,delimiter=',')
print(data)
print("\n\n")
actualData=np.array(data[1:,1:],dtype=int)
print(actualData)
```

### Slicing

```python
# data of last 5 years
print(data[:,6:])

# data of 2022
print(data[:,7])
```

### Operation: Sales Analysis


```python
# Total Sales each year

print("total sales each year:\n",np.sum(actualData,axis=0))


# average sales each year
print("\naverage sales each year:\n",np.mean(actualData,axis=0))


# maximum sale from each year
print("\nmaximum sale of each year \n",np.max(actualData,axis=0))


# minimum sale from each year
print("\nminimum sale of each year \n",np.min(actualData,axis=0))


bestMonths=list(str(data[i+1][0]) for i in np.argmax(actualData,axis=0))
print("\nbest month sales of each year:\n",bestMonths)


worstMonths=list(str(data[i+1][0]) for i in np.argmin(actualData,axis=0))
print("\nworst month sales of each year:\n",worstMonths)
```
